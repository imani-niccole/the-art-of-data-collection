A university is studying how different factors influence student success in an introductory programming course.

### Research Question: Does the AI tool improve performance?

* Statistical Test: T-Test

* This research question requires a T-Test because the outcome is a numerical outcome (final exam score), not simply a "Yes" or "No." Further, two groups means can be compared, students who passed/didn't pass and whether or not they used AI tools.  


```python
####Insert code here####
from scipy.stats import ttest_ind
group1 = df[df["study_group"] == "Control"]["score"]
group2 = df[df["study_group"] == "AI_Tool"]["score"]

stat, p = ttest_ind(group1, group2)
print(f"statistic: {stat:.3f}")
print(f"p-value: {p:.4f}")
```

### Research Question: Do outcomes differ across majors?

* Statistical Test: ANOVA

* Since 3+ groups are being analyzed (majors) and the outcome (final exam score) is a numerical value, ANOVA is the best statisitcal test to use for this question.

```python
####Insert code here####
from scipy.stats import f_oneway

groups = [
    df[df["study_group"] == s]["score"]
for s in df["study_group"].unique()
]
f_stat, p = f_oneway (*groups)
print(f"F={f_stat:.3f}, p = {p:.4f}")
```

### Research Question: Are certain groups more likely to pass?

* Statisitcal Test: Chi-Square Test

* The main reason why this research questions would not be a T-Test is because the two variables at hand are categorical, not numerical. The two variables are major and passed (Yes or No)

```python
####Insert code here####
from scipy.stats import chi2_contingency

# build a contingency table
table = pd.crosstab(
    df["study_group"],
    df["passed"]
)

chi2, p, dof, expected = (
    chi2_contingency(table)
)
print(f"chi2: {chi2:.3f}, p: {p:.4f}")
```

## Step 4: Interpretation

RQ 1: Does the AI tool improve performance?
Since the p-value is 0.4170 (41.7%) and is much higher than the alpha level of 0.05 (5%), we fail to reject the null hypothesis. This means there's a high chance that the outcomes are due to randomness rather than the AI tool itself. Plainly, the AI tool does not significantly improve exam performance. 



RQ 2: Do outcomes differ across majors? 
Interestingly, the p-value for the ANOVA test is the same as the T-test. Just the same, we fail to reject the null hypothesis. Since the p-value is so high, there's no significant difference in outcomes across the 3 majors. The F-Statistic (F = 0.664) represents the ration between the variation between group means and the variation within the groups. Since the F-value is much lower than 1, this signifies that the means of majors are grouped closer together rahter than spread apart. The limitation of ANOVA is that it tells you whether at least one group differents, not which one. Further testing must be used.



RQ 3: Are certain groups more likely to pass?
Again, for the Chi-test the p-value was 0.4550 which is much higher than the alpha value. This means there's a 45.5% probability that the difference between groups (AI tool and controlled) is due to chance. Further, we fail to reject the null hypothesis.
